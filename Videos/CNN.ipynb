```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense , Convolution2D , Flatten , AveragePooling2D
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train ,y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_train = X_train.astype("float32")[..., None] / 255.0
X_test = X_test.astype("float32")[..., None] / 255.0
```

## LeNet

```python
model  = Sequential()
model.add(Convolution2D(32 , kernel_size=(5,5),padding= 'valid', activation='tanh' , input_shape = (28,28,1)))
model.add(AveragePooling2D(pool_size=(2,2), strides = 2, padding = 'valid'))

model.add(Convolution2D(16 ,kernel_size=(5,5) , padding='valid' , activation='tanh'))
model.add(AveragePooling2D(pool_size=(2,2) , strides = 2 , padding='valid'))

model.add(Flatten())

model.add(Dense(120, activation='tanh'))
model.add(Dense(84, activation='tanh'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam' , loss = 'sparse_categorical_crossentropy' , metrics = ['accuracy']) 
```

```python
model.fit(X_train ,y_train, validation_data=(X_test, y_test) ,epochs = 5 )
```

```python
import numpy as np

y_prob = model.predict(X_test)         
y_pred = np.argmax(y_prob, axis=1)     
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test ,y_pred))
```
